```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/labeled_dataset.csv")
df.info()
df.head()
```

```python
sns.countplot(data=df, x="label")
plt.title("Follow-on funding: 0 = no follow-on, 1 = raised series-a+")
plt.show()
```

```python
top_categories = df["category_code"].value_counts().head(10).index
subset = df[df["category_code"].isin(top_categories)]

sns.barplot(data=subset, x="category_code", y="label", estimator="mean")
plt.xticks(rotation=45)
plt.ylabel("Follow-on rate")
plt.title("Follow-on success rate by sector (top 10)")
plt.show()
```

```python
sns.boxplot(data=df, x="label", y="angel_amount_usd")
plt.yscale("log")
plt.title("Angel round size vs follow-on outcome")
plt.show()
```
